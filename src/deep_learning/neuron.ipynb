#### ***Train Our Perceptron***

```python
import torch  
import torch.nn as nn 

# Inputs 
torch.manual_seed(42)
inputs = torch.tensor([1.2,2.0,3.2])

neuron = nn.Linear(in_features=3,out_features=1)
```

```python
output = neuron(inputs)
output
```

```python
# Verify Our Neuron 
neuron.weight @ inputs + neuron.bias
```

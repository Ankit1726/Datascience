#### ***Numpy For Data Science***

```python
import numpy as np 
arr = np.array([1,2,3,4,5])
print(arr,type(arr))
```

- ***Numpy Creation***

```python
# from list 
arr = np.array([1,2,3,"AIML","DS"])
print(arr,type(arr))
```

```python
# From 2D Matrix
arr = np.array([[1,2,3,4,5],[7,8,9,10,12]])
print(arr,arr.shape)
```

```python
# From 3D Matrix 
arr = np.array([[1,2,3],[4,5,6],[6,7,8]])
print(arr,arr.shape)
```

- ***Numpy Inbuilt Operations***

```python
arr = np.zeros((3,4))
print(arr,arr.shape)
```

```python
arr = np.ones((2,2))
print(arr,arr.shape)
```

```python
arr = np.full((3,3),8)
print(arr,arr.shape)
```

```python
arr = np.eye(3)
print(arr,arr.shape)
```

```python
# generating random number series 
arr = np.arange(1,20,3)
print(arr)
```

```python
arr = np.linspace(1,10,7)
print(arr)
```

```python
## Arrays Properties
arr = np.array([[1,2,3],
               [3,4,5],[7,8,9]])

print(arr.shape) # dimensions - (m x n)
print(arr.size) # total elements - m*n
print(arr.ndim) # dimension
print(arr.dtype) # data type object
```

```python
# Changing Datatype 
str_ar = np.array([1,2,3],dtype="U")
print(str_ar,str_ar.dtype)
```

```python
float_ar = np.array([1,2,3,4],dtype="float64")
print(float_ar,float_ar.dtype)
```

```python
int_ar = float_ar.astype(np.int32)
print(int_ar,int_ar.dtype)
```

```python
float_ar = int_ar.astype(np.float64)
print(float_ar,float_ar.dtype)
```

```python
# Array Reshaping 
ar = np.array([1,2,3,4,5,6])
print(ar.shape)

# convert 1D --> 2D
reshaped = ar.reshape((3,2))
print(reshaped,reshaped.shape)

# convert 2D --> 1D
flatten = reshaped.flatten()
print(flatten,flatten.shape)
```

```python
## Indexing
arr = np.array([1,2,3,4,5,7])
print(arr[2])
print(arr[0])
```

```python
## 2D Aray Indexing
arr = np.array([[1,2,3],[4,5,7],[2,8,9]])
print(arr[0][1])
print(arr[1][2])
```

```python
## Fancy Indexing
arr = np.array([1,2,3,4,5,7])
idx = [0,2,4]
print(arr[idx])
```

```python
## Boolean Indexing
print(arr[arr > 2])
print(arr[arr%2 == 0])
```

```python
arr = np.array([[1,3,2],[4,5,7],[8,9,10]])
print(arr)
print(np.sum(arr))

col_sum = np.sum(arr,axis=0)
print(col_sum)

row_sum = np.sum(arr,axis=1)
print(row_sum)
```

- ravel( ) → Returns a view (if possible), faster, changes may affect the original array.

- flatten() → Returns a copy, changes do not affect the original array.

```python
arr_2d = np.array([[1,2,3],[4,5,6]])
print(arr_2d.ravel())
print(arr_2d.flatten())
```

```python
arr_2d[0][2] = 7
print(arr_2d.ravel())
print(arr_2d.flatten())
```

```python
print(arr_2d)
```

```python
## Common Datatype
ar1 = np.array([1,2,3,4,5])
ar2 = np.array([1.0,2.0,5.0,7.0])
ar3 = np.array(["Python","DS","AIML"])

print(ar1.dtype)
print(ar2.dtype)
print(ar3.dtype)
print(ar1.nbytes)
```

```python
##  Complex Numbers
arr1 = np.array([2 + 3j])
arr2 = np.array([5 + 8j])
print(arr1, arr1.dtype)
```

```python
print(arr1+arr2)
print(arr1-arr2)
```

```python
## Vectorization
arr = np.array([1, 2, 3, 4, 5])

sq_arr = arr ** 2       
print(sq_arr)

arr2 = np.array([6, 7, 8, 9, 10])
print(arr + arr2)      
```

```python
#Broadcasting
arr_mul10 = arr * 10  
print(arr_mul10)

arr1D = np.array([1, 2, 3])
arr2D = np.array([[1, 2, 3], [4, 5, 6]])
print(arr1D + arr2D)
```

```python
## Vector Normalization
arr = np.array([[1,2,3],[4,5,7]])
mean = np.mean(arr)
std_dv = np.std(arr)

print((arr-mean)/std_dv)
```

```python
## Mathematical Functions
arr = np.array([1, 2, 3, 4, 5])

print(np.sum(arr))     
print(np.prod(arr))    
print(np.min(arr))    
print(np.argmin(arr))  
print(np.max(arr))     
print(np.argmax(arr))  
print(np.mean(arr))    
print(np.median(arr))  
print(np.std(arr))     
print(np.var(arr)) 
```

```python
arr = np.array([1,2,3,4,5,7])
print(np.square(arr))   
print(np.sqrt(arr))     
print(np.pow(arr, 3))
```

```python
print(np.log(arr))
print(np.log10(arr))
print(np.log2(arr))
print(np.exp(arr))
```

```python
print(np.round(2.678))    
print(np.floor(2.678))    
print(np.ceil(2.678))     
print(np.trunc(2.678))
```

```python
arr = np.array([1, 2, -5, 3, 8, -4, 2, 5])

print(np.abs(arr))     
print(np.sort(arr))  
print(np.unique(arr))
```

```python
# Vertical Stacking --> row wise & Horizontal Stacking --> column wise
ar1 = np.array([1,2,3,4,5])
ar2 = np.array([5,6,7,8,9])

vstack = np.vstack((ar1,ar2))
hstack = np.hstack((ar1,ar2))
```

```python
print(vstack)
print(hstack)
```

- ***Handling Missing Values With Numpy***

```python
arr = np.array([1,2,np.nan,6,9,np.nan])
print(arr)
print(np.isnan(arr))
```

```python
# Numeric Value
cleaned_arr = np.nan_to_num(arr,nan=0)
print(cleaned_arr)
```

```python
# Infinte Value
arr = np.array([1,2,3,np.inf,4,-np.inf,7])
print(arr)
print(np.isinf(arr))
```

```python
cleaned_arr = np.nan_to_num(arr,posinf=1,neginf=-1)
print(cleaned_arr)
```

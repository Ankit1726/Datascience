#### ***Pandas For Data Science***

```python
import numpy as np 
import pandas as pd 

df = pd.DataFrame({
     "name":["Ankit","Alex","Muskan","Mehak"],
     "cgpa":[8,7,8.5,9]
})
df
```

```python
type(df)
```

```python
s = pd.Series([23,34,45,78,88])
print(s)
print(type(s))
```

```python
s.index
```

```python
## custom indexing
series = pd.Series([23,44,57,89],index=["A","B","C","D"])
series
```

```python
## Vectorized operation
s1 = pd.Series([1,2,3,4])
s2 = pd.Series([7,8,9,10])
print(s1+s2)
```

```python
## Mutable Values but immutable size
s = pd.Series([1,2,3,4,57,8,9])
s[2] = 120
print(s)

changed_series = s.drop(1)
print(changed_series)
```

```python
## Datafram in pandas  using (Dictionary)
info = {
    "Name" : ["Adam", "Eve", "Bob"],
    "Marks" : [78, 99, 85],
    "Grade" : ['B', 'O', 'A']
}

df = pd.DataFrame(info)
df
```

```python
print(df.index)  # row labels
print(df.columns) # columns label
```

```python
## Dataframe in pandas  using (Numpy Array)
arr = np.array([[1,2,3],[4,5,7],[8,9,10]])
df = pd.DataFrame(arr,columns=['A','B','C'])
df
```

```python
## Dataframe in pandas  using (Lists)
lst = [["Ankit",99],["Kush",97],["Kavi",90],["Ankit",95]]
df = pd.DataFrame(lst,columns=["Name","Marks"])
df
```
